```python
ip = adv.drop('Sales',axis=1)
op = adv.Sales
```

```python
ip
```

```python
op
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(ip,op,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
```

```python
#Model Training
model = Sequential()
#Input Layer & First Hidden Layer
model.add(Dense(30, input_dim=3, activation='linear'))
#Hidden Layer
model.add(Dense(25, activation='linear'))
model.add(Dense(10, activation='linear'))
#Output Layer
model.add(Dense(1, activation='linear'))
model.compile(Adam(learning_rate=0.01), loss='mse')
```

```python
model.summary()
```

```python
#Linear - for continuous
#Sigmoid - for categorical
#TanH - for categorical
#ReLU - fro both
```

```python
model.fit(x_train,y_train,validation_data=(x_test, y_test),epochs=20)
```

```python
pred = model.predict(x_test)
pred
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,pred)
```
