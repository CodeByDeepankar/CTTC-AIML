## Tokenization
<h4>
    It is used to break document/text in smaller unit know as token. It is of Two types: 1) SentTokenizer and 2) WordTokenizer
</h4>

## Stemming
<h4>
    It is used to normalize the words into its base or root forms by removing suffixes.<br>
    eg:- Played - Play, Waited - Wait, Studying - Studi
</h4>

## Lemmatization
<h4>
    It is used to converts words into their meaningfull dictonary form, somehow it is similar to stemming but output of a lemmatization is a proper word<br>
  
</h4>

## POS tags
<h4>
    POS stands for parts of speach which is used to assign grammatical roles to the words<br>
    eg:- The(determinat) cat(noun) killed(verb) the(determinat) rat(noun)
</h4>

## NER 
<h4>
    It is a process of detecting named entity such as weather it is a person name, place, organization, location, quantity, etc.
</h4>

## Chunking
<h4>
    Picking individual piece of information and grouping them in a single phrase is know as chunking    
</h4>

# Vectorizer

### It is a technique used to convert text into numerical vectors show that mechine learning algorithim can process text data. It is of Two types : 1) count vertorizer, 2) TF-IDF vectorizer(TermFrequency-Inverse Document Frequency)

#### Count Vectorizer: Creates a vocabulary of all the unique words and counts how many times each word appers in a document

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
txt = ["I love Singing and I love Painting."]
x = cv.fit_transform(txt)
```

```python
x
```

```python
print(x)
```

```python
print(cv.vocabulary_)
```

```python
print(x.toarray())
```

```python
txt = ["I love Singing and I love Painting.", "I love ice-cream"]
y = cv.fit_transform(txt)
```

```python
y
```

```python
print(y)
```

```python
print(cv.vocabulary_)
```

```python
print(y.toarray())
```

## TF-IDF Vectorizer (Term Frequency - Inverse Document Frequency)

## TF - It is defind as the number of time a word apper in a document devided by total number of words in the document

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
txt = ["I love Singing and I love Painting.", "I love ice-cream"]
z = tfidf.fit_transform(txt)
```

## IDF - Inverse Document Frequency :- It is defind as the log of total number of documents devided by the number of documents that containt the word

```python
print(z)
```

```python
print(tfidf.vocabulary_)
```

```python
print(z.toarray())
```
