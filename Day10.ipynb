# MatplotLib - It is a data visualization technique in which is used to plot different graphs.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = [2, 4, 6, 8, 9, 5, 7]
y = [4, 3, 2, 5, 7, 8, 5]
plt.plot(x, y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [2, 4, 6, 8, 9, 5, 7]
y = [4, 3, 2, 5, 7, 8, 5]
plt.plot(x, y, 'o')
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [2, 4, 6, 8, 9, 5, 7]
y = [4, 3, 2, 5, 7, 8, 5]
plt.plot(x, y, 'o-')
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [2, 4, 6, 8, 9, 5, 7]
y = [4, 3, 2, 5, 7, 8, 5]
plt.plot(x, y, 'o--')
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [2, 4, 6, 8, 9, 5, 7]
y = [4, 3, 2, 5, 7, 8, 5]
plt.plot(x, y, 'ro--')
plt.title('Line Plot', color='b')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
# plt.show()
```

```python
x = [2, 4, 6, 8, 9, 5, 7]
y = [4, 3, 2, 5, 7, 8, 5]
plt.plot(x, y, 'or--')
plt.title('Line Plot', color='g')
plt.xlabel('X-axis', color='m')
plt.ylabel('Y-axis', color='c')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
x = [2,4,6,8]
y = [1,3,5,7]
plt.bar(x,y)
plt.show()
```

```python
x = [2,4,6,8]
y = [1,3,5,7]
plt.bar(x,y, color=['b', 'g', 'r', 'm'], width=0.9)
plt.title('Bar Plot', color='y')
plt.annotate("2026", xy=(1.8,1.2))
plt.show()
```

```python
x = [2,4,6,8]
y = [1,3,5,7]
plt.bar(x,y, color='g', width=0.8, label='Boys')
a = [1,3,5,7]
b = [2,4,6,7]
plt.bar(a,b, color='c', label='Girls')
plt.title('Bar Plot', color='y')
plt.annotate("2026", xy=(1.8,1.2))
plt.legend()
plt.show()
```
