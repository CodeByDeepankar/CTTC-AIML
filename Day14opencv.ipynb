```python
cap = cv2.VideoCapture(0)
if cap.isOpened():
    ret, frame = cap.read()
    print(ret)
    print(frame)
else:
    ret = False

image = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.title('AIML')
plt.axis('off')
plt.show()
cap.release()
```

```python
#Video
vdo = "vdo.mp4"
cv2.namedWindow(vdo)
cap = cv2.VideoCapture(vdo)
if cap.isOpened():
    ret, frame = cap.read()
    print(ret)
    print(frame)
else:
    ret = False

while ret:
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imshow(vdo,frame)
    if cv2.waitKey(10)==ord('q'):
        break
cv2.destroyAllWindows()
cap.release()
```
