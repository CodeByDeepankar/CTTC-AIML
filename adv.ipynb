### 5. ip/op Creation

```python
ip = adv.drop('Sales', axis=1)
op = adv.Sales
```

### 6. Train_Test_Split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(ip, op, test_size=0.2, random_state=3000)
```

```python
x_train
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

### z = x - µ / σ

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test  = sc.transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
x_train = mm.fit_transform(x_train)
x_test  = mm.transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
pred = lr.predict(x_test)
pred
```

```python
y_test
```

```python
 from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test, pred)
```

```python
r2 = r2_score(y_test, pred)
```

```python
print("MSE: ", mse)
print("R2 Score: ", r2)
```

```python
df = pd.DataFrame({'Y_test': list(y_test), 'Prediction': pred})
df
```

```python
sns.lmplot(x="Y_test", y="Prediction", data=df, scatter_kws={"color": "blue"})
plt.show()
```
