```python
cap = cv2.VideoCapture(0)
if cap.isOpened():
    ret, frame = cap.read()
    print(ret)
    print(frame)
else:
    ret = False

image = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.title('AIML')
plt.axis('off')
plt.show()
cap.release()
```

```python
#Video
vdo = "vdo.mp4"
cv2.namedWindow(vdo)
cap = cv2.VideoCapture(vdo)
if cap.isOpened():
    ret, frame = cap.read()
    print(ret)
    print(frame)
else:
    ret = False

while ret:
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imshow(vdo,frame)
    if cv2.waitKey(0)==ord('q'):
        break
cv2.destroyAllWindows()
cap.release()
```

```python
#Face Detection using Haarcascade
face = 'haarcascade_frontalface_default (1).xml'
model = cv2.CascadeClassifier(face)

img = cv2.imread("face2.jpg")

facepoints = model.detectMultiScale(img)
for x,y,w,h in facepoints:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),3)
print(facepoints)
plt.imshow(img[::,::,::-1])
plt.show()
```

```python
face = "haarcascade_frontalface_default (1).xml"
cap = cv2.VideoCapture(0)

ret, frame = cap.read()

while ret:
    ret, frame = cap.read()
    #print(ret)
    #print(frame)

    if not ret:
        break
    model = cv2.CascadeClassifier(face)
    facepoints = model.detectMultiScale(frame)

    for x,y,w,h in facepoints:
        cv2.rectangle(frame, (x,y), (x+w,y+h), (0,255,0), 3)

    cv2.imshow('Apna Face',frame)
    if cv2.waitKey(10)==ord('q'):
        break

cv2.destroyAllWindows()
cap.release()
```
