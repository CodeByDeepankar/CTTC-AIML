```python
#Importing Packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib
```

```python
#Reading and Studying the Dataset
iris = pd.read_csv("iris.data", header=None)
iris
```

```python
iris.columns = ['SL', 'SW', 'PL', 'PW', 'Flower Type']
iris
```

```python
iris.isnull().sum()
```

```python
iris.dtypes
```

```python
#Encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
iris['Flower Type'] = le.fit_transform(iris['Flower Type'])
iris
```

```python
iris['Flower Type'].unique()
```

```python
le.inverse_transform([0,1,2])
```

```python
ip=iris.drop('Flower Type',axis=1)
op=iris['Flower Type']
```

```python
from keras.utils import to_categorical
op=to_categorical(op, 3)
```

```python
op
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(ip,op,test_size=0.2,random_state=42)
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
```

```python
model = Sequential()
model.add(Dense(20, input_dim=4, activation='relu'))

model.add(Dense(35, activation='relu'))
model.add(Dense(15, activation='relu'))

model.add(Dense(3, activation='softmax'))
model.compile(Adam(learning_rate=0.01), loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,validation_data=(x_test,y_test),epochs=15)
```

```python
pred = model.predict(x_test)
pred
```

```python
p = np.argmax(pred, axis=1)
p
```

```python
y_test
```

```python
y = np.argmax(y_test, axis=1)
y
```

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score
ac = accuracy_score(y, p)
rc = recall_score(y, p, average='micro')
pc = precision_score(y, p, average='micro')
f1 = f1_score(y, p, average='micro')

print('Accuracy:', ac)
print('Recall:', rc)
print('Precision:', pc)
print('F1 Score:', f1)
```
