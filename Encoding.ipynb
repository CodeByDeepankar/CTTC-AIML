# Encoding

### It is used to convert string data into numeric values
###### Types of Encoding:- 
                       Label Encoding
                       One-Hot Encoding
                       Ordinal Encoding
                       Target Encoding
                       Frequency Encoding
                       Dummy Encoding

### Label Encoding: Replace each unique category with a unique number in a single column.
### One-Hot Encoding: Create a new column for each unique category and use 1 if the category is present, otherwise 0.
