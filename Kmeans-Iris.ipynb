```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = pd.read_csv('iris.data')
iris
```

```python
iris = pd.read_csv('iris.data', header=None)
iris
```

```python
iris.columns = ['SL', 'SW', 'PL', 'PW', 'Flower Type']
iris
```

```python
iris.isnull().sum()
```

```python
for i in iris.columns:
    print(iris[i].unique())
```

```python
for i in iris.columns:
    print(i, ':', '\n', iris[i].unique(), '\n')
```

```python
iris.describe(include='all')
```

```python
# we'll take the input only
ip = iris.drop('Flower Type', axis=1)
ip
```

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3)
km.fit(ip)
```

```python
km_pred = km.predict(ip)
km_pred
```

```python
centroid = km.cluster_centers_
centroid
```

```python
iris["Prediction"] = km_pred
iris
```

```python
sns.scatterplot(x=iris.SL, y=iris.PL, hue=iris['Flower Type'])
plt.show()
```

```python
sns.scatterplot(x=iris.SL, y=iris.PL, hue=iris['Prediction'])
plt.show()
```

```python
sns.scatterplot(x=iris.SL, y=iris.PL, c=km_pred)
sns.scatterplot(x=centroid[::,0], y=centroid[::,2], s=100, color='k')
plt.show()
```
