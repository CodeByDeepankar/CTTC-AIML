```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = pd.read_csv('iris.data', header=None)
iris
```

```python
iris.columns = ['SL', 'SW', 'PL', 'PW', 'Flower Type']
iris
```

```python
#distribution plot
sns.distplot(iris.SL)
plt.show()
```

```python
iris.SW.value_counts()
```

```python
sns.histplot(iris.SW)
plt.show()
```

```python
sns.displot(iris.SW)
plt.show()
```

```python
sns.kdeplot(iris.PW)
plt.show()
```

```python
sns.kdeplot(iris.PL, fill=True, color='gold', hatch='\|/', edgecolor='g')
plt.show()
```

```python
sns.histplot(iris.PW, color='orange', fill=False, hatch='\\*//')
plt.show()
```

```python
sns.distplot(iris.SW[iris['Flower Type']=='Iris-setosa'])
sns.distplot(iris.SW[iris['Flower Type']=='Iris-versicolor'])
sns.distplot(iris.SW[iris['Flower Type']=='Iris-virginica'])
plt.show()
```

```python
sns.displot(iris.SL[iris['Flower Type']=='Iris-setosa'])
sns.displot(iris.SL[iris['Flower Type']=='Iris-versicolor'])
sns.displot(iris.SL[iris['Flower Type']=='Iris-virginica'])
plt.show()
```

```python
sns.histplot(iris.SL[iris['Flower Type']=='Iris-setosa'])
sns.histplot(iris.SL[iris['Flower Type']=='Iris-versicolor'])
sns.histplot(iris.SL[iris['Flower Type']=='Iris-virginica'])
plt.show()
```

```python
sns.kdeplot(iris.SW[iris['Flower Type']=='Iris-setosa'], fill=True)
sns.kdeplot(iris.SW[iris['Flower Type']=='Iris-versicolor'], fill=True)
sns.kdeplot(iris.SW[iris['Flower Type']=='Iris-virginica'], fill=True)
plt.show()
```

```python
sns.boxplot(iris.SW)
plt.show()
```

```python
sns.boxplot(x=iris['Flower Type'], y=iris.SL)
plt.show()
```

```python
sns.boxplot(x=iris['Flower Type'], y=iris.SL, hue=iris['Flower Type'])
plt.show()
```

```python
#boxen plot
sns.boxenplot(x=iris['Flower Type'], y= iris.SL, palette='viridis')
plt.show()
```
