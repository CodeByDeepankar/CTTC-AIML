```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```



| Nature of Data  | Plots |
| -------- | -------- |
| Continuous v/s Continuous    | JointPlot, HeatMap, Scatter  |
| Continuous v/s Categorical    |   DistPlot, HistPlot, DisPlot, KdePlot, SwarmPlot, StripPlot, ViolinPlot, BoxPlot, BoxenPlot, PointPlot, BarPlot |
| Categorical    |   CountPlot |
| For all coulmns    |   PairPlot, JointPlot |
