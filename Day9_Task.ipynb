# Adult Dataset Cleaning

```python
import numpy as np
import pandas as pd
```

```python
adult = pd.read_csv('adult.data', header=None, skipinitialspace=True)
adult
```

```python
adult.columns = ['age', 'workclass', 'fnlwgt', 'education', 'education-num', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'capital_gain', 'capital_loss', 'hours_per_week', 'native_country', 'income']
adult
```

```python
adult.isnull().sum()
```

```python
adult.dtypes
```

```python
for i in adult.columns:
    print(i, ':', '\n', adult[i].unique(), '\n')
```

```python
adult.info()
```

```python
for i in adult.columns:
    print(i, ':', sum(adult[i]=='?'))
```

```python
d = adult.describe(include='all')
d
```

```python
adult.workclass.replace('?', 'Private', inplace=True)
adult.occupation.replace('?', 'Prof-specialty', inplace=True)
adult.native_country.replace('?', 'United-States', inplace=True)
```
