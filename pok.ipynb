```python
pok['Type 2'] = pok['Type 2'].fillna('None')
pok.isnull().sum()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(pok['Combat Power'], kde=True)

plt.title("Combat Power Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(pok['HP'], kde=True)

plt.title("HP Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=pok[['HP','Attack','Defense',
                      'Sp. Atk','Sp. Def','Speed']])

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    pok.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    y='Type 1',
    data=pok,
    order=pok['Type 1'].value_counts().index
)

plt.title("Pokemon Count by Type")

plt.show()
```

```python
sns.countplot(
    x='Legendary',
    data=pok
)

plt.title("Legendary Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    x='Type 1',
    y='Combat Power',
    data=pok
)

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x='Attack',
    y='Defense',
    hue='Type 1',
    data=pok
)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x='HP',
    y='Combat Power',
    data=pok
)

plt.title("HP vs Combat Power")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x='Speed',
    y='Combat Power',
    data=pok
)

plt.title("Speed vs Combat Power")

plt.show()
```

```python
sns.pairplot(
    pok[['Combat Power',
         'HP',
         'Attack',
         'Defense',
         'Sp. Atk',
         'Sp. Def',
         'Speed']]
)

plt.show()
```
